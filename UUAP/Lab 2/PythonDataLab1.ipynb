```python
monthly_consumption = [320, 300, 280, 350, 400, 420]
for value in monthly_consumption:
    print(value)
```

```python
total = 0
for value in monthly_consumption:
 total = total + value
print("Total consumption:", total)
```

```python
average = total / len(monthly_consumption)
print("Average consumption:", average)
```
